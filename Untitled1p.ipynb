```python
from datetime import date, timedelta


class Medicine:
    def __init__(self, med_id, name, category, price, quantity, expiry_date):
        self.med_id = med_id
        self.name = name
        self.category = category
        self.price = float(price)
        self.quantity = int(quantity)
        self.expiry_date = expiry_date  # a datetime.date object

    def is_expired(self):
        return self.expiry_date < date.today()

    def days_to_expiry(self):
        return (self.expiry_date - date.today()).days

    def status(self):
        if self.is_expired():
            return "EXPIRED"
        if self.days_to_expiry() <= 30:
            return f"Expires in {self.days_to_expiry()}d"
        return "OK"


class Pharmacy:
    LOW_STOCK_LIMIT = 15

    def __init__(self, name):
        self.name = name
        self.medicines = {}
        self.sales = []

    def add_medicine(self, med_id, name, category, price, quantity, expiry_date):
        if med_id in self.medicines:
            self.medicines[med_id].quantity += quantity
            print(f"[RESTOCK] {name}: +{quantity} units "
                  f"(now {self.medicines[med_id].quantity})")
        else:
            self.medicines[med_id] = Medicine(med_id, name, category, price, quantity, expiry_date)
            print(f"[ADDED]   {name} ({quantity} units)")

    def sell_medicine(self, med_id, quantity, customer="Walk-in"):
        med = self.medicines.get(med_id)
        if med is None:
            print(f"[FAILED]  Medicine ID {med_id} not found.")
            return
        if med.is_expired():
            print(f"[BLOCKED] {med.name} expired on {med.expiry_date}. Cannot sell.")
            return
        if med.quantity < quantity:
            print(f"[FAILED]  Only {med.quantity} units of {med.name} left "
                  f"(requested {quantity}).")
            return

        med.quantity -= quantity
        total = med.price * quantity
        self.sales.append({"date": date.today(), "customer": customer,
                           "name": med.name, "qty": quantity, "total": total})
        print(f"[SOLD]    {quantity} x {med.name} to {customer} = Rs.{total:.2f}")
        if med.quantity <= self.LOW_STOCK_LIMIT:
            print(f"          Low stock alert: {med.name} has {med.quantity} left!")

    def check_stock(self):
        print(f"\n{'ID':<6}{'Name':<22}{'Category':<14}{'Price':>8}{'Qty':>6}  {'Expiry':<12}Status")
        print("-" * 82)
        for m in self.medicines.values():
            print(f"{m.med_id:<6}{m.name:<22}{m.category:<14}{m.price:>8.2f}"
                  f"{m.quantity:>6}  {str(m.expiry_date):<12}{m.status()}")

    def low_stock_report(self):
        print("\n--- Low Stock Report ---")
        items = [m for m in self.medicines.values() if m.quantity <= self.LOW_STOCK_LIMIT]
        for m in items:
            print(f"{m.name}: {m.quantity} units left")
        if not items:
            print("All medicines are sufficiently stocked.")

    def expiry_report(self):
        print("\n--- Expiry Report ---")
        found = False
        for m in self.medicines.values():
            if m.status() != "OK":
                print(f"{m.name:<22} expiry {m.expiry_date}  -> {m.status()}")
                found = True
        if not found:
            print("No expired or soon-to-expire medicines.")

    def sales_report(self):
        print("\n--- Sales Report ---")
        for s in self.sales:
            print(f"{s['date']} | {s['customer']:<10} | {s['name']:<22} | "
                  f"Qty {s['qty']:<3} | Rs.{s['total']:.2f}")
        print(f"Total revenue: Rs.{sum(s['total'] for s in self.sales):.2f}")

        # Best seller
        totals = {}
        for s in self.sales:
            totals[s["name"]] = totals.get(s["name"], 0) + s["qty"]
        if totals:
            best = max(totals, key=totals.get)
            print(f"Best seller: {best} ({totals[best]} units)")


# ===================== IMAGINARY DATA & DEMO =====================
today = date.today()
pharmacy = Pharmacy("HealthPlus Pharmacy, Ghaziabad")

print("===== STEP 1: Adding medicines =====")
# (id, name, category, price, quantity, expiry date)
pharmacy.add_medicine("M001", "Paracetamol 500mg",  "Painkiller",  2.50, 200, today + timedelta(days=540))
pharmacy.add_medicine("M002", "Amoxicillin 250mg",  "Antibiotic",  12.00, 60, today + timedelta(days=300))
pharmacy.add_medicine("M003", "Cetirizine 10mg",    "Antiallergy",  3.00, 120, today + timedelta(days=20))
pharmacy.add_medicine("M004", "Cough Syrup 100ml",  "Syrup",       85.00, 25, today - timedelta(days=45))   # expired
pharmacy.add_medicine("M005", "ORS Sachet",         "Hydration",   18.00, 80, today + timedelta(days=400))
pharmacy.add_medicine("M006", "Pantoprazole 40mg",  "Antacid",      7.50, 14, today + timedelta(days=210))
pharmacy.add_medicine("M007", "Vitamin C 500mg",    "Supplement",   5.00, 150, today + timedelta(days=25))
pharmacy.add_medicine("M008", "Ibuprofen 400mg",    "Painkiller",   4.00, 90, today + timedelta(days=365))

print("\n===== STEP 2: Current stock =====")
pharmacy.check_stock()

print("\n===== STEP 3: Selling medicines =====")
pharmacy.sell_medicine("M001", 30, "Rahul Sharma")
pharmacy.sell_medicine("M002", 10, "Priya Verma")
pharmacy.sell_medicine("M003", 15, "Amit Singh")
pharmacy.sell_medicine("M004", 2,  "Neha Gupta")     # expired -> blocked
pharmacy.sell_medicine("M006", 5,  "Sunita Devi")    # triggers low stock alert
pharmacy.sell_medicine("M005", 100, "Karan Mehta")   # more than stock -> fails
pharmacy.sell_medicine("M005", 25, "Karan Mehta")
pharmacy.sell_medicine("M009", 1,  "Ravi Kumar")     # unknown ID

print("\n===== STEP 4: Restocking =====")
pharmacy.add_medicine("M006", "Pantoprazole 40mg", "Antacid", 7.50, 100, today + timedelta(days=210))

print("\n===== STEP 5: Updated stock =====")
pharmacy.check_stock()

print("\n===== STEP 6: Reports =====")
pharmacy.low_stock_report()
pharmacy.expiry_report()
pharmacy.sales_report()
```
